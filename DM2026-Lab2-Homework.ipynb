### 1.1 Preprocessing Steps

* **Merging:** The post text, the emotion labels and the train/test split are stored in three separate files. I merged them into one table by matching `post_id`.
* **Splitting:** Using the `set` column, the data is divided into 3,360 training posts (with labels) and 1,440 test posts (to predict).
* **Cleaning for TF-IDF:** Lowercase all text, remove URLs and @mentions, and squeeze extra spaces. In practice, lowercasing changed most posts, while the dataset contains no URLs and only one @mention.
* **Raw text for pretrained models:** For MiniLM and the emotion model, I used the original text instead of the cleaned version. These models were trained on natural sentences, so cleaning could remove useful signals (e.g. capital letters can express strong emotion).
* **Class imbalance:** The labels are unbalanced (875 posts for joy but only 288 for fear). Since macro F1 scores each of the six classes separately and then averages them, rare classes matter as much as common ones. Therefore, I used `class_weight="balanced"` in both LinearSVC and Logistic Regression, so mistakes on rare classes are penalized more.

### 1.2 Feature Engineering Steps

in this experiment i used three feature engineering methods: char_wb TF-IDF, MiniLM and emotion model + MiniLM.

in tfidf i used the method unlike the master notebook, which takes in words as wholes, but instead segment words into small parts, letting typos or variations of words have less effect on the understanding of the prediction.

in pretrained `all-MiniLM-L6-v2` model MiniLM, every document is turned into a 384-number vector by the MiniLM, and we would then hand the vector to logistic regression to learn the characteristics of number values of certain emotions.

The pretrained emotion model `j-hartmann/emotion-english-distilroberta-base` further calculates 7 emotion probabilities (6 required emotions plus one neutral) of a certain document, and the last method is to combine the emotion probability and the MiniLM-generated vector and leave it to logistic regression to learn.

Note: the MiniLM and emotion-model features are computed inside the Model section cells, next to the classifiers that use them.

### 1.3 Explanation of Your Model

* **Validation**
**80/20 split** as mentioned in the master notebook, it would be better if we leave some test cases beside the actual test case to test if the prediction is working, so i left about 20% of the 3,360 training posts (672 posts) for validation set.

stratify: to make sure that the ratio of each emotion in the validation set is the same as in the whole training set, lest a rare class like fear get too few examples and the validation score would be unreliable.

also, i set the random_state = 42 to make sure all tests are made on the same dataset to keep everything in control except the model type.

# **Competition Code**

## 1. Preprocessing Steps

```python
### Add the code related to the preprocessing steps in cells inside this section
import pandas as pd

# Load the three data files
posts = pd.read_json("data/kaggle/final_posts.json")          # post_id, text
labels = pd.read_csv("data/kaggle/emotion.csv")               # post_id, label (train only)
split = pd.read_csv("data/kaggle/data_identification.csv")    # post_id, set (train/test)

# Merge them into one table using post_id
df = posts.merge(split, on="post_id").merge(labels, on="post_id", how="left")

# Separate train (has labels) and test (no labels, to predict)
train_df = df[df["set"] == "train"].reset_index(drop=True)
test_df = df[df["set"] == "test"].reset_index(drop=True)

print("train:", train_df.shape, "| test:", test_df.shape)
print(train_df["label"].value_counts())
train_df.head()
```

```python
import re

# Four rules, applied to every tweet in order.
def clean_text(s: str) -> str:
    s = s.lower()                             # Scared, SCARED -> scared
    s = re.sub(r"http\S+|www\.\S+", " ", s)   # URLs
    s = re.sub(r"@\w+", " ", s)               # @mentions
    return re.sub(r"\s+", " ", s).strip()     # squeeze the gaps the rules left

# New column, so the original "text" column survives for human reading.
train_df["clean"] = train_df["text"].map(clean_text)
test_df["clean"] = test_df["text"].map(clean_text)

# Check the rules by eye: print two tweets before and after.
for i in [1, 2]:
    print("RAW:  ", train_df["text"].iloc[i])
    print("CLEAN:", train_df["clean"].iloc[i])
    print()
```

```python
# Check how much each cleaning rule actually changes the data (supports report 1.1)
all_text = pd.concat([train_df["text"], test_df["text"]])
print("posts with URLs:", all_text.str.contains(r"http\S+|www\.", regex=True).sum())
print("posts with @mentions:", all_text.str.contains(r"@\w+", regex=True).sum())
print("posts changed by lowercasing:", (all_text != all_text.str.lower()).sum())
```

## 2. Feature Engineering Steps

```python
#Character n-gram TF-IDF

### Add the code related to the feature engineering steps in cells inside this section
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 5), sublinear_tf=True)
X_train_tfidf = tfidf_vectorizer.fit_transform(train_df["clean"])
X_test_tfidf = tfidf_vectorizer.transform(test_df["clean"])

# what fraction of the matrix is empty; nnz is the number of non-zero entries
density = X_train_tfidf.nnz / (X_train_tfidf.shape[0] * X_train_tfidf.shape[1])
print(f"TF-IDF sparsity: {100 * (1 - density):.2f}% zeros")

# a window into the middle of the alphabetical vocabulary
print("some feature names:", tfidf_vectorizer.get_feature_names_out()[2000:2008])
```

## 3. Model Implementation Steps

```python
#TF-IDF + LinearSVC — validation
### Add the code related to the model implementation steps in cells inside this section
from sklearn.svm import LinearSVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score

y_train = train_df["label"]

# Hold out 20% of the training posts as a validation set
X_tr, X_val, y_tr, y_val = train_test_split(
    X_train_tfidf, y_train, test_size=0.2, random_state=42, stratify=y_train)

clf = LinearSVC(class_weight="balanced", C=0.5, random_state=42).fit(X_tr, y_tr)
print("validation macro F1:", round(f1_score(y_val, clf.predict(X_val), average="macro"), 4))
```

```python
# submission format

final_clf = LinearSVC(class_weight="balanced", C=0.5, random_state=42).fit(X_train_tfidf, y_train)
test_pred = final_clf.predict(X_test_tfidf)
submission = pd.DataFrame({"id": test_df["post_id"], "label": test_pred})
submission.to_csv("submission.csv", index=False)
```

```python
# trying MiniLM

from sentence_transformers import SentenceTransformer
from sklearn.linear_model import LogisticRegression

st_model = SentenceTransformer("all-MiniLM-L6-v2")
E_train = st_model.encode(train_df["text"].tolist(), show_progress_bar=True)
E_test = st_model.encode(test_df["text"].tolist(), show_progress_bar=True)

E_tr, E_val, y_tr2, y_val2 = train_test_split(
    E_train, y_train, test_size=0.2, random_state=42, stratify=y_train)

lr = LogisticRegression(max_iter=2000, class_weight="balanced").fit(E_tr, y_tr2)
print("MiniLM validation macro F1:", round(f1_score(y_val2, lr.predict(E_val), average="macro"), 4))
```

```python
#Emotion-model probabilities + MiniLM
import numpy as np
from transformers import pipeline

#open source model
emo = pipeline("text-classification",
               model="j-hartmann/emotion-english-distilroberta-base",
               top_k=None, truncation=True)

def emo_probs(texts):
    out = emo(texts, batch_size=32)
    # emotion probability 
    return np.array([[d["score"] for d in sorted(o, key=lambda d: d["label"])] for o in out])

P_train = emo_probs(train_df["text"].tolist())
P_test = emo_probs(test_df["text"].tolist())

# emotion + MiniLM 
F_train = np.hstack([P_train, E_train])
F_test = np.hstack([P_test, E_test])

F_tr, F_val, y_tr3, y_val3 = train_test_split(
    F_train, y_train, test_size=0.2, random_state=42, stratify=y_train)

lr2 = LogisticRegression(max_iter=2000, class_weight="balanced").fit(F_tr, y_tr3)
print("Emotion+MiniLM validation macro F1:",
      round(f1_score(y_val3, lr2.predict(F_val), average="macro"), 4))
```

```python
#Fine-tune emotion model
import torch, copy
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from sklearn.utils.class_weight import compute_class_weight

device = "mps" if torch.backends.mps.is_available() else "cpu"
print("device:", device)

name = "j-hartmann/emotion-english-distilroberta-base"
labels = sorted(train_df["label"].unique())
l2i = {l: i for i, l in enumerate(labels)}
y_all = np.array([l2i[l] for l in train_df["label"]])

idx = np.arange(len(train_df))
tr_idx, val_idx = train_test_split(idx, test_size=0.2, random_state=42, stratify=y_train)

tok = AutoTokenizer.from_pretrained(name)
model = AutoModelForSequenceClassification.from_pretrained(
    name, num_labels=len(labels), ignore_mismatched_sizes=True).to(device)

def encode(texts):
    e = tok(texts, truncation=True, padding=True, max_length=128, return_tensors="pt")
    return e["input_ids"], e["attention_mask"]

texts_all = train_df["text"].tolist()
w = torch.tensor(compute_class_weight("balanced", classes=np.arange(len(labels)), y=y_all),
                 dtype=torch.float).to(device)
loss_fn = torch.nn.CrossEntropyLoss(weight=w)
opt = torch.optim.AdamW(model.parameters(), lr=2e-5)

def predict(texts, bs=64):
    model.eval(); preds = []
    with torch.no_grad():
        for i in range(0, len(texts), bs):
            ids, mask = encode(texts[i:i+bs])
            logits = model(input_ids=ids.to(device), attention_mask=mask.to(device)).logits
            preds.append(logits.argmax(-1).cpu().numpy())
    return np.concatenate(preds)

#run for three epochs, choose the best one

best_f1, best_state = 0, None
val_texts = [texts_all[i] for i in val_idx]
for epoch in range(3):
    model.train()
    perm = np.random.RandomState(epoch).permutation(tr_idx)
    for i in range(0, len(perm), 16):
        b = perm[i:i+16]
        ids, mask = encode([texts_all[j] for j in b])
        logits = model(input_ids=ids.to(device), attention_mask=mask.to(device)).logits
        loss = loss_fn(logits, torch.tensor(y_all[b]).to(device))
        opt.zero_grad(); loss.backward(); opt.step()
    f1 = f1_score(y_all[val_idx], predict(val_texts), average="macro")
    print(f"epoch {epoch+1}  validation macro F1: {f1:.4f}")
    if f1 > best_f1:
        best_f1, best_state = f1, copy.deepcopy(model.state_dict())

model.load_state_dict(best_state)
test_pred3 = [labels[i] for i in predict(test_df["text"].tolist())]
pd.DataFrame({"id": test_df["post_id"], "label": test_pred3}).to_csv("submission_ft.csv", index=False)
print("best validation F1:", round(best_f1, 4))
```

```python
# # Emotion + MiniLM: train on all data, write submission_emo.csv
final_lr = LogisticRegression(max_iter=2000, class_weight="balanced").fit(F_train, y_train)
test_pred2 = final_lr.predict(F_test)
sub2 = pd.DataFrame({"id": test_df["post_id"], "label": test_pred2})
sub2.to_csv("submission_emo.csv", index=False)
```

```python
import torch.nn.functional as Fnn

def predict_proba(texts, bs=64):
    model.eval(); out = []
    with torch.no_grad():
        for i in range(0, len(texts), bs):
            ids, mask = encode(texts[i:i+bs])
            logits = model(input_ids=ids.to(device), attention_mask=mask.to(device)).logits
            out.append(Fnn.softmax(logits, -1).cpu().numpy())
    return np.concatenate(out)

#ensemble: a mixture of the previous two approaches
P_ft_val = predict_proba(val_texts)
P_lr_val = lr2.predict_proba(F_val)          # lr2.classes_ is also alphabetical
ens_val = [labels[i] for i in ((P_ft_val + P_lr_val) / 2).argmax(1)]
print("Ensemble validation macro F1:", round(f1_score(y_val3, ens_val, average="macro"), 4))

#take the average
P_ft_test = predict_proba(test_df["text"].tolist())
P_lr_test = final_lr.predict_proba(F_test)
ens_test = [labels[i] for i in ((P_ft_test + P_lr_test) / 2).argmax(1)]
pd.DataFrame({"id": test_df["post_id"], "label": ens_test}).to_csv("submission_ens.csv", index=False)
```
